```python
import pandas as pd


data = {
    "product": ["Phone", "Laptop", "Mouse", "Keyboard"],
    "price": [800, 1500, 30, 70],
    "quantity": [3, 2, 10, 5]
}

df = pd.DataFrame(data)


df["total"] = df["price"] * df["quantity"]

print("DataFrame:\n")
print(df)


high_total = df[df["total"] > 300]

print("\nProducts with total > 300:\n")
print(high_total)


average_price = df["price"].mean()

print("\nAverage price:", average_price)


sorted_df = df.sort_values(by="total", ascending=False)

print("\nSorted by total (descending):\n")
print(sorted_df)
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr"]
sales = [120, 180, 150, 220]


plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

for i in range(len(months)):
    plt.text(i, sales[i], str(sales[i]), ha='center', va='bottom')

plt.show()
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

iris=load_iris()

x=iris.data
y=iris.target
X_train, X_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)


model = LogisticRegression(max_iter=200)

model.fit(X_train, y_train)


y_pred = model.predict(X_test)


accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)

new_sample = [[5.1, 3.5, 1.4, 0.2]]

prediction = model.predict(new_sample)

print("Predicted class:", prediction[0])
print("Predicted flower:", iris.target_names[prediction[0]])
```
